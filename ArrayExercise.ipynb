```python
import numpy as np
```

```python
#Columns: [Age, Math Marks, Science Marks]
data = np.array([[18, 85, 78],
                [19, 92, 88], 
                [17, 76, 95], 
                [18, 65, 70], 
                [20, 90, 85]])
data
```

```python
#Get the shape of the matrix.
MatrixShape = data.shape
MatrixShape
```

```python
#Find the average age of students.
avgAgeMean = data[:,0].mean()
avgAgeMean
```

```python
#Extract Math marks of all students.
mathmarks = data[:,1]
mathmarks
```

```python
#Find the highest Science mark.
higsciMarks = data[:,2].max()
higsciMarks
```

```python
#Get details of the student who scored more than 90 in Math.
Morethan90 = data[:,1] > 90
Morethan90
```

```python
#Increase Math marks of all students by 5.
markinc = data[:,1] + 5
markinc
```

```python
#Find how many students are younger than 19.
young = data[:,0] < 19
young
```

```python
#Calculate the average marks in each subject (column-wise mean).abs
avgm = data[:,1:].mean(axis=0)
avgm
```

```python
#Get data of students who scored at least 80 in both subjects.
both = data[(data[:,1] >= 80) & (data[:,2] >= 80)]
both
```
